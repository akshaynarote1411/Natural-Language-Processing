```python
import re
```

```python
pattern = r"hello"
str1 = "hello world"

pattern in str1
# 'hello' in ' hello world' => yes
```

```python
result = re.match(pattern,str1)
result
result.group()
```

```python
pattern = r"hello"
str1 = "world hello"
result = re.match(pattern,str1)
result 

# here the 'pattern' and 'str1' will not match by using re.match() because "hello" is not in begninning of both the string
# here instead of re.match(), we should use re.search()
```

Note:
    
    - re.match will checks for at biginning of string 
    
    - if we want to identify anywhere then we can go for re.search

```python
pattern = r"hello"
str1 = "world hello"
result = re.search(pattern,str1)
print(result)
result.group()
```

```python
pattern = r"\d+"
str1 = "the price is 100 and tax is 20"
result = re.findall(pattern,str1)
print("result:",result)
```

```python
pattern = r"\d+"
str1 = "order number 1234 shipped on 28th sept. 2024"
result = re.sub(pattern,'omkar',str1)
print("result:",result)
```

```python
pattern = r"^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}$"   # '^' this symbol is used to start the email pattern(means it denote that email start from there)
                                                              # '$' this symbol is used to end the email pattern
email="omkar.nallagoni@gmail.com"
bool(re.match(pattern,email))
```

```python
import re
pattern = '\d{3}'   # d{3} it will used to split the numbers in groups of 3 numbers.
str1 = 'my mobile number is 1234567890'
re.findall(pattern,str1)
```

```python
import re
pattern = '\d{2}'   # d{2} it will used to split the numbers in groups of 2 numbers.
str1 = 'my mobile number is 1234567890'
re.findall(pattern,str1)
```

```python
import re
pattern = '\d{4}'   # d{4} it will used to split the numbers in groups of 4 numbers.
str1 = 'my mobile number is 1234567890'
re.findall(pattern,str1)
```

```python
str1 = "my number is 123-456-7890"
pattern = '\d{3}-\d{3}-\d{4}'
re.findall(pattern,str1)
```

```python
str1 = "my number is (123)-456-7890"
pattern = '\(\d{3}\)-\d{3}-\d{4}'
re.findall(pattern,str1)
```

```python
str1 = "my number is (123) 456-7890"
pattern = '\(\d{3}\) \d{3}-\d{4}'
re.findall(pattern,str1)
```
