```python
doc1 = "going rain today."
doc2 = "Today going outside."
doc3 = "going watch season premiere."
Doc = [doc1, doc2, doc3]
print(Doc)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(Doc)
X
```
